with open()의 자원 자동 해제

```python
class MyResource: 
    def __enter__(self):
        print("__enter__호출")
        return self
     #끝날떄, 예외발생시 호출됨.
     #esc는 예외 발생시 값이 생김.
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__호출")



```

```python
with MyResource() as o: #__enter__호출, 결과는 o라는 참조변수
    print(o)
```

```python
#파일에서 작성하는 위치(파일포인터) 가장 앞으로 이동해서 내용배치. 
# 기존것 지워짐
with open("text.txt","w",encoding="UTF-8") as fp:
    fp.write("hello!")
```

```python
#파일포인터 이동하지 않고 누적해서 작성
with open("text.txt","a",encoding="UTF-8") as fp:
    fp.write("\n 안녕하세요")
```

텍스트 파일 읽기 함수 (read, readline,readlines)

```python
with open ('text.txt', "r", encoding = "UTF-8") as fp:
    text = fp.read()
    print(text)
```

```python
with open ('text.txt', "r", encoding = "UTF-8") as fp:
    text = fp.readlines()
    print(text)
```

```python
with open ('text.txt', "r", encoding = "UTF-8") as fp:
    while(text := fp.readline()):
        print(text)
```

```python
with open ('text.txt', "r", encoding = "UTF-8") as fp:
    lines = fp.readlines()
    print(lines)
```

```python
with open ('text.txt', "r", encoding = "UTF-8") as fp:
    for line in fp:
        print(line)
```

```python
import csv

with open ("memers.csv","w", encoding = "UTF=8", newline="") as fp:
    writer = csv.writer(fp)

    writer.writerow(["name", "age","mobile"])
    writer.writerow(["justine",20,"010-0000-0000"])
```

```python
with open ("memers.csv", "r", encoding = "UTF=8") as fp:
    reader = csv.reader(fp)
    members = list(reader)

    print(members)
```

JSON형식

```python
import json #표준라이브러리 불러오기

#dict ->json파일로 변환
user = {
    "name" :"justine",
    "age":20,
    "mobile":"010-0000-0000"
}

with open ("user.json","w", encoding="UTF=8") as fp:
    json.dump(user,fp, indent=4,ensure_ascii=False) #각항목마다 네 칸씩 띄어쓰기
```

```python
with open("user.json","r",encoding="UTF=8") as fp:
          user2=json.load(fp)

          print(user2, "type:", type(user2))
```
