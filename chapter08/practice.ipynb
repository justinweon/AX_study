```python
class Hero:
    def __init__(self, name): #name = instance variable
        self.name = name
        self.hp = 100 #default setting

    #define instance method 객체가 만들어지면 접근할수 있는 변수/매서드.
    def take_damage(self, damage):
        #self.hp -= damage #possible, but it may push out negative
        #if hp is positive, take out the damage and pushes it out 
        
        minus_damage = damage if self.hp >= damage else self.hp #3항조건. true이면 damage, false이면self.hp

        self.hp -= minus_damage
        print("남은체력:", self.hp)

        if self.hp <=0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")

```

```python
hero = Hero("Pikachu")
hero.take_damage(90)
hero.take_damage(20)
```

```python
class Doorlock:
    def __init__(self,password):
        if not self.validate(password):
            return
        
        self.__password = str(password)

    def get_password(self):
        return self.__password

    def set_password(self, password):
        
        if not self.validate(password):
            return
        self.__password = str(password)

    def validate(self,password):

        result = len(str(password)) >= 4  #result = true
        if not result:
            print("경고: 비밀번호는 4자리 이상이어야 합니다.")

        return result 
        
```

```python
door = Doorlock(234565)
```

```python
print(door.get_password())
```

```python
class User:
    def __init__ (self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init(name)
        self.point = point 

```

```python
user = VIPUser("james", 25)

print(user.name, user.point)
```
