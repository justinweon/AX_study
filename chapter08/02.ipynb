#### 상속 inheritence

```python
class Animal: 
    def move(self):
        print('동물이 움직인다.')
    def __init__(self, kind):
        self.kind = kind #동물종류


class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") #부모쪽 인스턴스변수인 kind에 초기화가 됨. 

    def move(self):
        super().move() #부모쪽 객체의 move()를 호출
        print (self.kind + "강아지 걸음으로 움직인다.")


class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print (self.kind+"고양이 걸음으로 움직인다.")
```

```python
print(dog.kind, cat.kind)
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
class schedule:
    def __init__(self, year, month, day):
        self.year=year
        self.month=month
        self.day= day

    def __str__(self):
        return f"year = {self.year}, month = {self.month}, day = {self.day}"
```

```python
s1 = schedule(2026,8,18)
print(s1)
```

```python
#통제. 참조로 값이 대입되기 때문에 의도치 않은 값이 할당될 수 있음. 어플리케이션 오류 가능성 있음.
#지금은 통제 불가능.
#통제불가능하면 직접접근을 막아야한다! 변수의 접근을 막아주는 파이썬 기법. (__) <- 외부조건 안들음.
#인스턴스 변수를 외부에서 접근하지 못하게 만들어야한다. 


```

```python
class schedule:
    def __init__(self, year, month, day):
        self.__year=year
        self.__month=month
        self.__day= day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"

```

```python
s1= schedule(2026,8,18)
print(s1)
```

```python
s1.__year = 2026
s1.__month = 2
s1.__day = 31

print(s1)
```

```python
class schedule:
    def __init__(self, year, month, day):
        self.__year=year
        self.__month=month
        self.__day= day

    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month
    
    def get_day(self):
        return self.__day

    def set_year(self,year):
        self.__year = year

    def set_month(self,month):
        self.__month = month

    def set_day(self,day):
        if self.__month ==2 and day >28: #값에대한 통제가 가능
            day = 28
        self.__day = day
        

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = schedule(2026,8,18)

s1.set_month(2)
s1.set_day(31)

print(s1)

#변수를 대입할때는 바로 들어가서 통제가 불가능. 지금은 한가지 가능성이있음. 값을 설정하는게 set 매서드이기 때문. set앞에 코드를 대입할수있음.함수로 정의되어있으면 통제가능.
```
