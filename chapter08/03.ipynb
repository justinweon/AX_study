```python
class Calculator:
    count = 0

    #clz -> calculator 클래스 객체 참조 변수
    #clz -> 클래스 변수, 다른 클래스 접근 목적.

    @classmethod
    def add(clz, num1, num2):
        clz.count +=1
        print("호출 횟수:", clz.count)
        return num1 +num2
    
    @staticmethod
    def minus(num1, num2):
        return num1 - num2
```

```python
Calculator.add(10,20)
```
