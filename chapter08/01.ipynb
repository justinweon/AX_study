#### class and instance

```python
class Bus: 

    #속성정의
    def __init__(self): #첫번째 매개변수는 거의 대부분 self. 역할이 정해져있음
        self.count = 0 #승객수, 초기화: 변수가 되었을때 최초로 답을 대입
        self.passenger = []

    def ride(self, passenger): #버스탑승기능 
        self.count += 1 #승객수 증가
        self.passenger.append(passenger)
        print("탑승")

    def off(self): #하차
        if self.count > 1:
            self.count -= 1 #승객수 감소
            self.passenger.remove(self.passenger)
        print("하차")

    def show_count(self):
        print("승객수:",self.count)

    def print_self(self):
        print(id(self))


class Passenger: 
    pass


#설계도면. 객체를 정의하는데 쓰기 떄문에 매서드라고 불리기도 함. 매서드: 객체에 소속되어있는 함수.
```

#### __str__

```python
class Bus: 

    
    def __init__(self): 
        self.count = 0 
        self.passenger = []

    def ride(self, passenger): 
        self.count += 1 
        self.passenger.append(passenger)
        print("탑승")

    def off(self): #하차
        if self.count > 1:
            self.count -= 1 
            self.passenger.remove(self.passenger)
        print("하차")

    def show_count(self):
        print("승객수:",self.count)

    def print_self(self):
        print(id(self))
    
    def __str__(self):
        return f"승객수: {self.count}, 승객: {self.passenger}" #문자열로 출력하기때문에 f-string


class Passenger: 
    def __init__(self, name, age): #호출할떄 투입되는 변수. 객체를 생성하기 위해 매개변수로 정의된 지역변수
        self.name = name #self.name/self.age 는 인스턴스변수. 객체가 생성되면 만들어지는 인스턴스 변수
        self.age= age

    def __str__(self):
        return f"이름: {self.name}, 나이 {self.age}"
```

```python
bus = Bus()
p1 = Passenger()
p2 = Passenger()

bus.ride(p1)
bus.ride(p2)

print(bus)

# __str__로 정의 
```

```python
bus = Bus()
p1 = Passenger("김철수", 20)
p2 = Passenger("김영희", 25)

bus.ride(p1)
bus.ride(p2)

print(bus)
```

#### __repr__

프린트를 하면 자동으로 호출되는 메서드. __str__과 비슷 
__str__가 우선순위가 높음. __st__이 먼저 호출됨. 
로깅(기록)목적, 디버깅용이라 우선순위가 높지 않음. 클래스명(매개변수=값, ...) 형태 -> eval () 함수로 문자열 실행햇을때 객체가 만들어지는 형태로 정의

eval(문자열) : 문자열을 파아ㅣ썬 코드로 해성해서 실행해줌/ 보안상 위험할 수 있음으로 사용에 유의.

```python
class Bus: 

    
    def __init__(self): 
        self.count = 0 
        self.passenger = []

    def ride(self, passenger): 
        self.count += 1 
        self.passenger.append(passenger)
        print("탑승")

    def off(self): #하차
        if self.count > 1:
            self.count -= 1 
            self.passenger.remove(self.passenger)
        print("하차")

    def show_count(self):
        print("승객수:",self.count)

    def print_self(self):
        print(id(self))
    
    def __str__(self):
        return f"승객수: {self.count}, 승객: {self.passenger}" #문자열로 출력하기때문에 f-string


class Passenger: 
    def __init__(self, name, age): #호출할떄 투입되는 변수. 객체를 생성하기 위해 매개변수로 정의된 지역변수
        self.name = name #self.name/self.age 는 인스턴스변수. 객체가 생성되면 만들어지는 인스턴스 변수
        self.age= age

    # def __str__(self):
    #     return f"이름: {self.name}, 나이 {self.age}"

    
    def __repr__(self):
        return f"Passenger name= {self.name}", age= {self.age}""
```

```python
bus = Bus()
p1 = Passenger("김철수", 20)
p2 = Passenger("김영희", 25)

bus.ride(p1)
bus.ride(p2)

print(bus)
```

```python
p1  = Passenger("김철수", 20)

print(p1) #__str__호출될것
print(repr(p1)) #__repr 호출될것
```
