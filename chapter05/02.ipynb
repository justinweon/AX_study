#### 변수 스코프

변수 스코프(scope): 변수를 접근할 수 있는 범위.
- 지역 변수
- 전역 변수 

콜 스택은 재료를 구입할수있는 상점. 콜스택안에서 함수가 호출될때 정의되는 값을 변수들은 호출될때만 변수가되고 유효함니다.
함수안에서만 유효하기 떄문에 지역변수라고 부름

함수밖에있는 변수들은 전역변수라고 불러요.모든 경루(파일내)에서 다 접근가능한 변수. 

함수는 범위가 좁고 밖은 다 가능함. 이것을 변수 스코프라고 합니다.

```python
a="global"


def outer():
    a = "enclosing"
    print("outer:",a)

    def inner():
        #a = "local" 제거
        global a #a라는 전역변수를 찾겠다 의미
        print("inner:",a)

    inner()

outer()

#전역변수를 취하고 싶은데 LEGB때문에 안될떄, 키워드 global를 사용해줌. 
#글로벌은 유의해서 써야함 부작용이있음. 전역변수가 수정되는 경우가 많아서.
```

```python


a= True


def outer():
    a = "enclosing"
    print("outer:",a)

    def inner():
        global a
        a = not a
        print("inner:",a)

    inner()

outer()

#아워함수가 중요한 값인데 계속 호출하면? 의도치않은 경우가 발생하기도 한다.? 외부의 값을 바꾸기 때문. 
#내부에만 있는 함수를 순수함수라고 함. 
```

```python
# 텍스트 메시지 저장소 구현하기

def message_store():
    storage = ""

    def add_message(message):
        nonlocal storage
        if storage:
            storage+="\n"+ message
        else:
            storage += message
        return storage
    return add_message
```

```python
my_storage = message_store()
```

```python
my_storage("새로운 알림이 도착했습니다.")
```

```python
# 데코레이터를 활용한 알림 테두리 장식하기

def deco_border(func):
    def wrapper():
        print("=====")
        func()
        print("=====")
    return wrapper


@deco_border
def show_welcome():
    print("환영합니다 스파르타님")

```

```python
#callback practice

def  send_sms(message):
    print(f"SMS notification: {message}")

def send_email(message):
    print(f"email notification: {message}")


def process_order(item,callback):
    print(f"processing order for {item}...")
    callback(f"your {item} has shipped!")


process_order("chicken", send_sms)
process_order("beef", send_email)
```

```python
def add(a,b):
    return a+b #sum(a,b) <??? 왜 안대. 리스트 구조로 넣어서 사용 가능. 연산자사용이 더 좋음! 더 빠르고 가볍다.
def multiply(a,b):
    return a*b 

def calculate(a, b, callback):
    result = callback(a,b)
    print(f"result:{result}")


calculate(10,5,add)
calculate(10,5,multiply)
```
