#### 함수의 기본 정의

#### 정의 방법

정의 = define 하지만 짧게, def = "정의할거야!"
함수를 저의할때는 반드시 앞에 def를 붙힘, 그리고 함수 상자에 이름을 지어야함. def 함수이름.
문을 여는것, 입력값 넣는 입구 -> ()

연산을 하려면 기록을 해야하잖아, 값을 전달하려면 공간이 필요하다, 메모리. -> 변수가 필요하다느느 말. 담아서 가져가. 함수상자에 들어가서 처리해. 
값을 대신 전달해주는 (매개하주는) 변수, 입구에 정의 되어야함. 함수의 재료를 받을 수 있는 변수 (매개변수)

입력 -> 배개변수를 정의 (없을 수도 있음)

입력이 들어오면, 처리를 해야함

출력은 별도의 키워드로 명시하면 출력을 담당함. 결과를 내보냄. 하단에 return

세가지 형태구성으로 함수가 만들어짐. 



```
def 함수이름 (매개변수, ...):
    #처리부분 정의

    return 반환값
```

#### 가변인수와 키워드 가변인수

*args = 위치 인수 기반 가변적 위치인, 매개변수 갯수 상관없이 지정 가능, 키워드 매개변수로는 사용 불가.
**kwargs = ** 두개 붙히면 됩니다.

````python
def add(num1, num2, num3):
    return num1+num2+num3
add(10,20,30)
````

````python
def add(num1, num2, num3, num4):
    return num1+num2+num3+num4
add(10,20,30,40)
````

````python
#가변 매개 변수 
#매개변수의 숫자에 상관없이 가변적으로 실행하고 싶을떄 앞에* 를 넣으면 가변적 매개변수라는 뜻

def add(*nums):
    print(nums, type(nums))
````

````python
add (10,20)
add (10,20,30)
add(10,20,30,40)
````

````python
def add(*nums):
    total = 0
    for num in nums:
        total += num
    return total 

#키워드 인수에서는 사용 x 변수명이 없기 때문.
````

````python
print(add(10,20))
print(add(10,20,30,))
print(add(10,20,30,40))
````

````python
#**kwargs 실습

def print_user_info(**info): #괄호 안에는 아무거나 적어도됨kwarg 가능
    print(info)
 
````

````python
print_user_info(name="정린", age = 20)

#딕셔너리로 출력
````

````python
#키워드는 위치인수보다 뒤에 위치. 문법. 
#키워드 매개변수는 가변적이라 하더라도 뒤에 위치합니다. 
````

````python
def print_user_info(name, age, **info):
    print("name:", name)
    print("age:", age)
    print(info)
````

````python
print_user_info("Justine", 20)
````

````python
print_user_info("Justine", 20, mobile = "010-1111-2222")
````

````python
def print_user_info(name, age, *args, **kwargs): #별표 두개는 별표 하나보다 꼭 뒤에 있어야 한다. 
    print("name:", name)
    print("age:", age)
    print(args)
    print(kwargs)
````

````python
print_user_info("Justine", 20, mobile = "010-1111-2222")

#가변인수는 필수가 아니라 비워져있음. 
````

````python
print_user_info("Justine", 20,'A', 20, mobile = "010-1111-2222") #A와 20은 args에 들어감, 튜플로 묶임.
````

````python
import sys

def outer_func():
    x = 10
    
    def inner_func():
        y = 20
        
        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()
        
        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")
        
        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
````

````python
# 배송비 자동 계산 및 다중 데이터 반환

def calculate_total(price,delivery_fee=3000):
    total = price + delivery_fee
    return (total,"원 결제가 완료되었습니다.")
````

````python
#무제한 과목 점수 합계 및 평균 산출

def get_total_and_average(*arg):
    if len(arg)==0:
        print(0,0)

    return sum(arg)/len(arg)

````
