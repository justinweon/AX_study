#### 함수의 기본 정의

#### 정의 방법

정의 = define 하지만 짧게, def = "정의할거야!"
함수를 저의할때는 반드시 앞에 def를 붙힘, 그리고 함수 상자에 이름을 지어야함. def 함수이름.
문을 여는것, 입력값 넣는 입구 -> ()

연산을 하려면 기록을 해야하잖아, 값을 전달하려면 공간이 필요하다, 메모리. -> 변수가 필요하다느느 말. 담아서 가져가. 함수상자에 들어가서 처리해. 
값을 대신 전달해주는 (매개하주는) 변수, 입구에 정의 되어야함. 함수의 재료를 받을 수 있는 변수 (매개변수)

입력 -> 배개변수를 정의 (없을 수도 있음)

입력이 들어오면, 처리를 해야함

출력은 별도의 키워드로 명시하면 출력을 담당함. 결과를 내보냄. 하단에 return

세가지 형태구성으로 함수가 만들어짐. 



```
def 함수이름 (매개변수, ...):
    #처리부분 정의

    return 반환값
```

#### 함수를 호출/실행하는 방법

#### 목적
함수는 왜 사용되나? 투입되는 값과 상관없이 수행되는 로직이 같기 떄문.
코드를 여러개 투입되는 값마다 만들 필요 없이 함수를 만들어놓고 사용할수 있음.

재활용이 목적

함수는 레시피같은거예요. 레시피에는 재료도 있고, 절차도나와있음. 정의는 정의일 뿐. 어떤 재료가 필요한지, 절차는 어떻게 만들어야 하는지
떡볶이 만들때 필요한것. 


````python
#함수이름을 짓는 규칙도 변수짓는 규칙과 같음, 가급적 snake_case
# calc = calculation

def calc(x):
    y = x*2+1

    return y

````

````python
#함수를 생성했으니 호출해보자!
#함수이름을 부르고 변수를 넣어 실행

result = calc(10)
result
````
