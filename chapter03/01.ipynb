#### 시퀀스 자료구조 

순서가 있는 데이터 군집
항상 위치를 계산해주는 연산자가 있음
    - 인덱스 연산자 : [순서번호], 데이터의 위치를 계산. 0번부터 시작. 0번이 가장 첫번쨰 순서. 
    - 리스트 : 내용 변경 가능. 가변 자료형
    - 튜플 : 내용 변경 불가능. 불변 자료형


#### 리스트

- 변경가능 
- [값1, 값2, ...] #대괄호에 감싸서 리스트 정리

#### 주요 매서드 (주요 기능)
1) 추가 (create)
    - append(x) : 끝에 추가.  -> lst.append(4), 원본 리스트 끝에 추가한다. 주소 동일.
    - insert(추가 위치, 추가할 요소) : 해당위치에 추가됨. 
    - extend (collection - 리스트, 튜플) : 한꺼번에 추가
    - + [값1, 값2...] : 플러스 기호로 추가 가능. 추가하고 새로운 리스트를 생성. 새로운 리스트에 대한 새로운 주소부여.
    
2) 조회 (read)
    - 인덱스 번호로 접근 -> fruit[1]
    - index(요소) : 특정 요소 인덱스 번호를 확인

3) 수정 (update)
    - fruits[1] = '오렌지' 인덱스 번호로 접근 한후 다른값을 대입

4) 삭제 (delete)
    - remove(요소) : 첫번째 발견한 특정 요소 제거
    - clear() : 리스트 모든 요소 제거

5) 기타
    - sort() : 정렬(기본값 오름차순, 알파벳순, 숫자 작은 순서대로)
    - sort(reverse=true) : 내림차순
    - reverse : 순서반전.
    - count(요소) : 특정요소의 등장 횟수

```python
fruits = ['apple', 'orange', 'banana']
```

```python
fruit1 = fruits[1]

fruit1

#인덱스를 통해 해당 요소 접근
#READ조회
```

```python
#해당 인덱스 값을 교체하면 수정이 된다

fruits[1] = '오렌지'

fruits
```

```python
fruits.append("melon")
fruits
```

```python
#요소 다른 변수에 복사

fruits_copy=fruits
fruits_copy
```

```python
#오렌지를 다시 영어로 바꿔볼까?

fruits_copy = fruits
fruits_copy[1] = "orange"
fruits_copy
```

```python
#요소 같이 수정됨.
#메모리에 리스트가 있는 상태. 주소있음. fruits_copy에 주소값을 부여. 동일한 값을 가르킴.참조.
#동일한 대상을 참조 하고 있는 것. 

print(fruits)
print(fruits_copy)
```

```python
fruits.extend(['cherry', 'peach', 'grapes'])
fruits
```

```python
# +기호로 추가가능

fruits = ["apple"]
fruits + ["orange", "peach"]
```

```python
#추가 전후 주소값 비교
#.extend 를 사용했을때는 주소가 같음.

fruits_extend = ["apple"]
print("추가 전(extend):", id(fruits_extend))

fruits_extend.extend(["orange", "banana"])
print("추가 후(extend):", id(fruits_extend))
```

```python
#+연산자를 사용했을때는 주소가 바뀜. 
#새로운것을 만드느냐 기존것을 수정하느냐 차이. 아주중요함. 

fruits = ["apple"]
print("추가 전(+):", id(fruits))

fruits = fruits + ["orange", "banana"]
print("추가 후(+):", id(fruits))
```

```python
#sort매서드 

fruits.sort()
fruits
```

```python
#sort 내림차순

fruits.reverse()
fruits
```

#### 튜플 (tuple)

변경 불가 리스트
리스트와 다르게 정의할때는, (값1, 값2, ...) 
append, remove 없음. 생성되면 변경이 안되기 때문에 추가 삭제 수정 전부 안됨. 
읽기전용 리스트.

#### 매서드 (주요 기능)
- 인덱스 연산 [] : 요소 접근
- COUNT(요소) : 요소 등장 횟수
- INDEX(요소) : 요소의 위치 번호 
- (값1,) : 값이 하나인 튜플은 반드시 , 를 함께 추가 

소괄호를 생렬하고 콤마로 값을 나열해서 정의 할 수도 있음.
