순차자ㄹ료가 아닌 경우 인덱스 함수 사용 x
사전자료구조 key-value (키워드 중복 허용x) 검색을 위한 목적 

#### 딕셔너리 자료구조

- key-value 쌍
- key : value 를 찾기 위한 키워드, 중복 x, 리스트는 사용 불가, 튜플은 가능. 
- value : 키로 찾은 검색 결과. 값. 중복 o
- {"key":value}   
- {"key":value, "key" : value ...}   

#### 주요기능
- 추가 (CREATE) 
    - 없는 키 이름으로 값을 대입. 
- 조회(READ)
    - 키값으로 조회. 연산자와 키 이름으로 조회. ['키 이름']
     - keys() -> 모든 키를 담은 view. 읽기전용
     - values() -> 모든 값을 담은 view. 읽기전용
     - items() -> (키, 값) 목록을 담은 view
- 

```python
#회원인증 예시

user = {
    "email":"test@test.org",
    "password": 1234,
    "name":"김철수"
    }

```

```python
#추가/조회 실습
#대괄호로 접근

user['mobile'] = '010-1000-1000'

user
```

```python
#조회는 인덱스 연산자로 

user['name']
```

```python
#views(읽기전용) 실습

user.keys()
```

```python
user.values()
```

```python
user.items()
```
