사전자료구조 key-value (키워드 중복 허용x) 검색을 위한 목적 

#### 딕셔너리 자료구조

- key-value 쌍
- key : value 를 찾기 위한 키워드, 중복 x, 리스트는 사용 불가, 튜플은 가능. 
- value : 키로 찾은 검색 결과. 값. 중복 o
- {"key":value}   
- {"key":value, "key" : value ...}   

#### 주요기능
- 추가 (CREATE) 
    - 없는 키 이름으로 값을 대입. 
- 조회(READ)
    - 키값으로 조회. 연산자와 키 이름으로 조회. ['키 이름'] - 키가 없으면 KeyError 발생
        - get("키 이름") - 키가 없으면 none 반환. 
     - keys() -> 모든 키를 담은 view. 읽기전용
     - values() -> 모든 값을 담은 view. 읽기전용
     - items() -> (키, 값) 목록을 담은 view
- 수정 (update)
    - 이미 있는 키값에 대입하면 수정됨. 
    - update({key, value, key, value ...})  -> 딕셔너리
- 삭제 (delete)
    - del items['key']
    - clear() -> 전체 비우기

- 다른 키-값 쌍 -> 딕셔너리로 변환. dict(...)

#### 집합 (set) 자료구조

- 중복 허용하지 않는다
- 순서가 결정되지 않는다

집합도 중괄호. 키가 없는 집합. 

정의 방법 -> 중괄호에 값을 넣으면 됨 {값1, 값2, 값3 ...}

#### 집합연산
연산자/매서드 
- 합집합: a | b  -> a.union(b)
- 교집합: a & b -> a.intersection(b) 
- 차집합: a - b -> a.difference(b)
- 대칭차집합: a ^ b -> a.symmetric_difference(b)

#### 집합 자료로 형변환
- set(리스트/튜플)



```python
#10이 세번이나 중복! 중복이 없기 때문에 10은 한번만 집합. 

a = {10, 10, 10, 20, 30,}

a
```

```python
#합집합, 교집합, 차집합, 대칭차집합

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8, 9}

print("합집합:", a|b)
print("교집합:", a&b)
print("차집합:", a-b)
print("대칭차집합", a^b)
```
