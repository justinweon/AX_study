#### 단일 조건문 if

조건문에서는 판별이 필요핫고 T/F이 필요함. 판별값이 나올수 있는 연산이 필요함. (비교)
파이썬에서는 빈 문자열, 0, none 은 거짓으로 판단됨
조건문에선 판별식이 필요하다! 

- 판별식에 따라서 참이면(T) 참에 실행되는 코드를 정의하고, 거짓이면 (F) 거짓조건에 실행되는 코드를 정의할 수있는 문장! 
- 판별식 : 비교연산자, 논리 연산자, 값을 가지고 판단 가능. 이런것으로 구성되어 있음.

`````python
#if뒤에 판별식, 참인지 거짓인지 판별할수 있는 식. 비교/물리 연산자!
#판별식 뒤네 콜론 기호가 옴. 
#규칙: 들여쓰기 규칙. '안쪽 조건에 해당한다' 라는 의미. tab키로 들여쓰기

```
if 판별식:
     판별식이 참일때 실행되는 코드 
````
`````

`````python
#실습 
#참 조건 확인

number = 10 

if number > 5 :
    print("5보다 크다")
`````

#### if-else 조건문

`````python
#5보다 크지 않은 경우
#if-else 조건문 
`````

`````python
num = 5

if num == 10:
    print("10 입니다")


#이 경우 거짓일때 아무것도 출력되지 않음. 참일 경우에만 실행되기 때문.
`````

`````python
# 판별식이 거짓일때 실행, else구문으로 가능. 
# 새로운 문장이 시작되는것을 알려줘야함. if와 같은 선상에 명령


num = 5

if num == 10:
    print("10 입니다")
else: 
    print("10이 아닙니다")
`````

#### 다중 조건 분기ㅣ if-elif-else

`````python
#input함수. 입력하는 것. 

age = input()
print(age)
`````

`````python
#문자형으로 사용한다.(?)

age = int(input())
print(age, type(age))
`````

`````python
#다중조건분기이긴 하지만 앞선조건이 참이면 앞 조건에서만 실행되고 이후로 넘어가지 않음
#논리이슈 발생.

age = int(input())

if age > 8:
    print("초등학생")
elif age >= 14:
    print("중학생")
`````

`````python
#초등학생과 중학생 자리를 바꿔주면 논리이슈 해소
#어디가 먼저 참이 되는지 확인후 하위로 배치

age = int(input())

if age > 14:
    print("중학생")
elif age >= 8:
    print("초등학생")
`````

`````python
#5는 안나옴. 앞선 조건문/판별식이 모두거짓일 시 else 실행됨.

age = int(input())
if age>= 20:
    print("성인")
elif age>=17:
    print("고등학생")
elif age >= 14:
    print("중학생")
elif age >= 8:
    print("초등학생")
else:
    print("유치원생")
`````

#### 복잡한 조건 처리: 중첩조건문과 논리 연산자 결합

`````python
#입력받은 숫자가 10이상 100미만인지 확인
#앞선 조건 참, 이휴 조건 참, if 안에 if를 넣어 중첩조건
#간단한 조건은 and/or연산자로 단순화 하는 것을 추천.

num = int(input())

if num >= 10:
    if num < 100:
        print(f"{num}는 10이상 100미만입니다.")
`````

`````python
#논리연산자 and 결합을 해서 단순화 해보자!

num = int(input())

if num >= 10 and num <100:
    print(f"{num}은 10이상 100미만입니다.")
`````
