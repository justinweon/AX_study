#### For 문

````python
#횟수가 정해져 있는 반복 (함수:range)

```
for num in range(#반복횟수) : 
    #반복 시행되는 코드 정의 
```

- range(반복횟수), 0~반복횟수
- range(start,end): 시작 이상 종료미만 출력
- range(start,end,증감단위)
- range()숫자 생성시에도 사용 
````

````python
#1부터 100까지 더하기 

#변수 정의부터 
total = 0
start = 1

for num in range(100):
    #total = total + start 가능하긴하나 변수가 반복이 되니까 간단히 복합대입을 사용.
    total += start
    start += 1

print(f"1~100까지 합계는 {total}입니다")
````

````python
#num 변수는 i라고도 쓰임.
#i : 순서를 의미하는 index의 앞자를 따서 나온 변수명
#개발자들 사이에서는 보통 i라는 변수명을 사용. i라고 사용함.
#for 문에서는, 중첩에서 i는 첫번째 조건, 그 다음은 알파벳 순서대로. i, j, k, L, M, N . . . (관례)
#변수명이 중요하지 않은 경우 자리만 채우는 언더바 (_)

total = 0
start = 1

for num in range(100):
    #total = total + start 가능하긴하나 변수가 반복이 되니까 간단히 복합대입을 사용.
    total += start
    start += 1
    print("반복횟수:", num)

print(f"1~100까지 합계는 {total}입니다")
````

````python
#반복횟수를 사용하지 않는 경우 언더바로 자리만 차지. 변수명이긴 하지만 관례적으로 그냥 자리만 차지하는 변수명. 
#반복횟수는 나오지않음.

for _ in range(10):
    print("반복")
````

````python
#과일, 홀수번째 과일만 필요한 경우 
#앞에 0은 생략. range는 시작점과 종료지점(포함x)

fruits = [
    "apple",
    "peach",
    "watermelon",
    "pear",
    "orange",
    "mango",
    "cherry",
    "guava",
    "melon",
    "apricot"
]

for i in range(1,5): #시작지점, 종료지점 정할수 있음.
    print(fruits[i])
````

````python


fruits = [
    "apple",
    "peach",
    "watermelon",
    "pear",
    "orange",
    "mango",
    "cherry",
    "guava",
    "melon",
    "apricot"
]

#range(,,)<< 콤마임
for i in range(0,10,2): #홀수 번째 과일만 가져오고 싶을때. 2씩 증가. 0, 2, 4, . . . >> 순서는 1, 3, 5, 홀수 번쨰가 됨. 
    print(fruits[i])
````

````python
#range()숫자 생성시에도 사용 
#* ->

nums = [*range(1,101)] #1부터 101까지 생성됨.
print(nums[:5]) #5만 보기위해 슬라이싱 사용
````
