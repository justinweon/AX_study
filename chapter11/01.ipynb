sync vs async

```python
# 동기식 함수 

import time

def work(title, seconds):
    time.sleep(seconds)
    print(title, "완료!")

start = time.perf_counter() #작업시작시간

work("작업A", 2)
work("작업B", 3)
work("작업C", 1)

end = time.perf_counter() #작업 종료시간 
print("걸린시간:", end - start)
```

```python
#비동기함수

import asyncio
import time

#async 넣어야만 비동기
async def async_work(title,seconds):
    await asyncio.sleep(seconds)
    print(title, "완료")

async def main():
    start = time.perf_counter()
    await asyncio.gather(
         async_work("작업A", 2),
         async_work("작업B", 3),
         async_work("작업C", 1)
    )

    end = time.perf_counter()
    print("걸린시간:", end - start)

await main()
```
