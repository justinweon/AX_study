```python
import asyncio

async def work(title, seconds):
    await asyncio.sleep(seconds)
    return title + " complete"
```

```python
async def main():
    await work ("task1", 2)
    await work ("task2", 3)
    await work ("taks3", 1)
```

```python
async def main():
    results = await asyncio.gather(work("task1",2), work ("task2", 3), work("taks3", 1))
    print(results)
```

```python
await main()
```

```python
asyncio.gather -> 동시에 실행하고 마지막 작업까지 대기했다가 한꺼번에 결과를 반환
```
