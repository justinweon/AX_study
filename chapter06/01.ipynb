```python
def cafe_order():
    yield "1번 손님 음료"
    yield "2번 손님 음료"

order = cafe_order()
```

```python
print(next(order))
```

```python
#yield 기반의 제너레이터 함수 작성

def cheer():
    yield "대박"
    yield "사건"
    yield "성공"

go = cheer()
```

```python
print(next(go))
```

```python
#연속된 정수의 제곱값 생성시 메모리 용량 측정
#list comprehension 과 제너레이터 표현식의 메모리 사용량을 측정하여 지연평가(lazy Evaluation)


import sys 

square_list = [i**2 for i in range(1,50001)]
gen = (i**2 for i in range(1,50001))
square_gen = gen

print(sys.getsizeof(square_list))
print(sys.getsizeof(square_gen))


```
