```python
persons = ["김철수","김영희","김민수"]
fruits = ["사과","오렌지","멜론"] #쌍을 지어서, 튜플형태로 내보내보자

```

```python
#간단하게한다면 list comprehension 가능 밑에처럼
items = []
for i, person in enumerate(persons):
    items.append((person,fruits[i]))

items
```

```python
#세줄 쓰기 싫다면 zip함수

items = zip(persons, fruits)
list(items)
```
