#### 비교연산자

<,>,>=,<=, 
== (같다) (*동등성, 동일한 가치를 가지고 있는지)
!= (다르다)

- 연산의 결과는 무조건 T,F (불리언형)
- 판별이 필요한 구문 (조건문, 반복문)에서 필요

- 동등성 vs 동일성
-- 동등성: 동일한 가치를 비교 (==)
-- 동일성: 동일한 대상인지 비교 (is)(주소비교)

#### 논리연산자 (and, or, not)

and
- 각 항이 T->T (하나라도F가 나오면 F)

or
- 가가 항의 하나라도 T->T 

not
- T조건을 F조건으로 반전, F조건을 T조건으로 반전

비교연산자보다 우선순위가 낮다. 그래야 연산이 정상적으로 동작. 각항의 연산이 끝난 다음 and or 진행 


```python
# and
num1=20
result1 =num1>10
result2=num1<100

result1 and result2
```

```python
# not (부정연산)

not(result1 and result2)
```

```python
fruit1="apple"
fruit2="pear"

result1=fruit1=="apple"
result2=fruit2=="orange"

result1, result2
```

```python
result1 or result2
```

```python
fruit1=="apple"or fruit2=="orange"
```

```python
result = True or 10**2

result

#판별목적의 연산. OR 연산은 한 항이 T이면 최종 결과가 T이기 때문에 판별(True)이 끝나면 이후 연산은 무시함
#첫 항의 결과 true = 판별 완료. 결과결정됨.
```

```python
result = False or 10**2

result

#판별목적의 연산. 판별(true)이 끝나지 않았기 때문에 이후 연산 진행후 출력
#첫 항의 결과 False = 판별 아직 결정되지 않음. 
```

```python
result = False and 10**2

result

# and 연산은 한 상이 F이면 최종 결과가 F이기 때문에, 판별 (false)이 끝나면 이후 연산 무시함.
```

```python
result = True and 10**2

result

#판별 (False)이 나지 않았기 떄문에 이후 연산 진행
```
