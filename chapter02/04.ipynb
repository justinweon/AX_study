#### 형식화 (formatting)

나의 이름은 이고 나이는 이다.  - 틀을 만들어 놓고 내용은 변경해가며 활용



1. 1세대: % 연산자

- &s = 문자열 대체
    - &숫자s -> 최소 숫자 자리수 만큼 공간 확보, 남는 공간은 공백
- &d = 정수 대체
    - &대체숫자d -> %05s ->남는 공백은 0으로 대체
- &f = 실수 대체
    - %.2f -> 소수점 두자리까지 표시
    - %10.2 -> 정수부 8자리, 소수점 자리 2까지

#문자열= 개발언어에서 문자는 문자 하나 (a, b). 문자열은 열지어있는 문자들 (apple)

```python
message = "my name is %s, and I am %d years old." #&s,&d(digit)= 대체가 되는 부분

#대체될 값들을 나열. 

message % ("James", 20)
```

```python
message = "my name is %10s, and I am %d years old." #%숫자s = 최소 숫자 자리수 만큼 공간 확보, 남는 공간은 공백

message % ("James", 20)
```

```python
message = "주문번호: %05d" % 10 #&대체숫자s -> %05s ->남는 공백은 0으로 대체
message
```

```python
message = "숫자: %f" % 10.123456789
message
```

```python
message = "숫자: %.2f" % 10.123456789
message
```

```python
message = "숫자: %10.2f" % 10.123456789
message
```

```python
message = "숫자: %010.2f" % 10.123456789
message
```

```python
message = "number: %d" % 10.12345 #정수만 출력
message
```

```python
message = "number: %s" % 10.12345 #잘 모르겠고 다 출력해야겠다면 문자(s) - %s
message
```

2. 2세대: str.formate() 매서드 방식

```python
#순서 지정 교체 방식은 보통 중간에 숫자를 넣음 -> {0} {1} {2} {3} 
#순서는 0부터 시작

message = "내 이름은 {0} 이고, 나이는 {1} 입니다."
message.format("철수", 20)

#정의한 순서대로가 아니라, 안에 있는 번호가 순서
```

```python
#순서를 바꾸면 바뀜

message = "내 이름은 {1} 이고, 나이는 {0} 입니다."
message.format("철수", 20)
```

```python
#단어 지정 교체는 단어를 찾아서 교체 -> {단어} {단어}

message = "내 이름은 {name}이고, 나이는 {age} 입니다."
message.format(name= "철수", age=20)
```

```python
#두 교체 방식 혼합 예시
#교체할수 있는 문자를 숫자를 먼저 넣어야 함
message = "내 이름은 {name}이고, 나이는 {0} 입니다."
message.format(20, name="철수")
```

3. 3세대: f-string방식

```python
#편리해진 방식

name = "영희"
age = 20 

message = f"내 이름은 {name}이고, 나이는 {age}입니다."

message
```

```python
#기능이 여러가지 추가됨.
#%d 최소공백 맞추는 기능인데 왼쪽 오른쪽 방향도 정해서 진행가능.
#예시

price = 10000
f"가격: {price}"
```

```python
#cntd
#천 단위로 표기

price = 10000
f"가격: {price:,}"
```

```python
#cntd
#자리수 지정 가능. 위치도 지정 가능.
#우측은 >, 좌측은< 하면됨.
#10 = 열자리. 남은 자리는 공백으로 채움

price = 10000
f"가격: {price:>10}원"
```

```python
price = 10000
f"가격: {price:<10}원"
```

```python
#디버깅 조절자 

f"{price=}"

```
