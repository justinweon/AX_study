2. 명시적 형변환 (explicit type conversion)

```python
연산은 같은 자료형끼리 가능. 파이썬은 명시적으로 자료형을 바꾸는 함수 제공 

형변환 함수
- int(..) -> 정수로 변환
- float(..) ->실수로 변환
- bool(...) ->true,false로 변환
- str(...)-> 문자형으로 변환 

강제 형변환
```

```python
num ="100"
type(num)  = str이 나올것 

int(num) +200 = 300 이 나올것
```

```python
"현재 시간은" + str(10) + "시"
```
