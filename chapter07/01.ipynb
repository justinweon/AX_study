```python
def login(user_id, user_pass):
    if not (user_id == "user01"):
        print("아이디가 일치하지 않습니다.")
        return
    
    if not (user_pass =="1234"):
        print("비밀번호가 일치하지 않습니다")
        return

    print("로그인 처리 로직...")
```

```python
def login(user_id, user_pass):
    if user_id != "user01":
            raise ValueError("아이디가 일치하지 않습니다.") #예외발생, 코드 실행 중단.
    

    if user_pass != "1234":
          raise ValueError ("비밀번호가일치하지 않습니다.")

    print("로그인 처리 로직...")
```

```python
#회원페이지

try:    
    login("user02","1234")

except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
```

```python
#관리자 페이지

try:    
    login("user02","1234")

except ValueError as e:
    print(e)
    print("관리자 페이지에서 추가적으로 할 로직...")
```

```python
class UserIdMismatchError(Exception):
    def __str__(self):
        return"아이디가 일치하지 않습니다"

class UserPassMismatchError(Exception):
    def __str__(self):
        return"비밀번호가 일치하지 않습니다"


#예외클래스가 되었음
```

```python
def login(user_id, user_pass):
    if user_id != "user01":
            raise UserIdMismatchError() #예외발생, 코드 실행 중단.
    

    if user_pass != "1234":
          raise UserPassMismatchError ()

    print("로그인 처리 로직...")
```

```python
#맛집메뉴 조회 프로그램 

matzip = [" ","떡볶이","순대","튀김"]

###메뉴판###
#1번 : 떡볶이
#2번 : 순대
#3번 : 튀김

try:
    choice = int(input())
    selected_menu = matzip[choice]

except ValueError:
    print("번호로 입력해주세요")

except IndexError:
    print("없는 번호입니다")

else:
    print("주문하신 메뉴:",selected_menu)

finally:
    print("감사합니다.")
```

```python
#### 상품 주문 수량 검증 프로그램

def order(order_count):
    if order_count <= 0:
        raise ValueError("주문 수량은 1개 이상이어야 합니다.")

    else:
        print(f"{order_count}개 주문 완료되었습니다.")

try:
    order(int(input()))

except Exception as e:
    print(f"[시스템 경고] {e}")

```

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```
